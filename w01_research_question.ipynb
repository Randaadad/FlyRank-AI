```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset shape:", df.shape)
print("Available columns:")
print(df.columns.tolist())
```

## 2. The question: decision, action, cost of a wrong call

## The question

### Research question

Which visible pages have lower-than-expected CTR or engagement compared with pages in the same position tier, and therefore deserve review first?

### Decision

The decision is which pages should be prioritized for CTR or engagement review.

### Action

A reviewer could investigate the page's title and meta description, search intent match, snippet structure, or on-page engagement. Some pages may simply need monitoring rather than an immediate change.

### Cost of a wrong recommendation

A false positive could waste a reviewer's time on a page that does not have a meaningful opportunity. A false negative could cause a potentially valuable page to be missed. Therefore, the system should support human review rather than automatically decide that a page needs to be changed.

```python
print("Research question defined.")
print("Decision: prioritize pages for CTR/engagement review.")
print("Action: human review of possible CTR/engagement opportunities.")
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains a large number of content pages with search and engagement measurements. The available signals include impressions, CTR, average position, sessions, and engagement metrics.

The following numbers are calculated directly from the starter dataset and help show whether CTR/engagement opportunity scoring is worth investigating.

```python
print("Number of rows:", len(df))
print("Average CTR:", round(df["ctr"].mean(), 4))
print("Average impressions (90d):", round(df["impressions_90d"].mean(), 2))

corr = df["avg_position"].corr(df["ctr"])
print("Correlation between avg_position and CTR:", round(corr, 3))
```

## 4. Careful words: what I can and can't claim

## Careful words: what I can and can't claim

This project will provide observational and directional findings from the available search and engagement data.

I can identify patterns and associations between CTR, position, impressions, engagement, and other observable signals. I can use these patterns to rank pages that may deserve human review.

The final output will be decision-support: a ranked list of opportunities for human review, supported by measurable evidence.

```python
print("Claims are observational and directional.")
print("The output is intended for decision-support and human review.")
print("No causal claims or claims about Google's algorithm will be made.")
```
